```python
import pandas as pd
import folium
from folium.plugins import HeatMap

df = pd.read_csv('log_dataset.csv')

customers = df[df['Type'] == 'Customer']

m = folium.Map(location=[10.77, 106.70], zoom_start=11)

heat_data = customers[['Latitude', 'Longitude']].values.tolist()

HeatMap(heat_data).add_to(m)

m
```
