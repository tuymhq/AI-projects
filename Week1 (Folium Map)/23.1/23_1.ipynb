```python
import folium

ueh_b = [10.7613, 106.6683]

places = [
    ([10.7580648, 106.6595238], 'Bệnh viện Chợ Rẫy'),
    ([10.7707539, 106.6698906], 'Vạn Hạnh Mall'),
    ([10.7654509, 106.6797695], 'Công viên Lý Thái Tổ'),
    ([10.7642682, 106.6825759], 'NowZone'),
    ([10.7825913, 106.6957684], 'Hồ Con Rùa')
]

m = folium.Map(location=ueh_b, zoom_start=14)

folium.Marker(
    location=ueh_b,
    popup='UEH - Cơ sở B',
    tooltip='Trường của tôi',
    icon=folium.Icon(color='hotpink', icon='graduation-cap', prefix='fa')
).add_to(m)

group = folium.FeatureGroup(name="Địa điểm gần UEH")

for loc, label in places:
    folium.Marker(
        location=loc,
        popup=label,
        tooltip=label,
        icon=folium.Icon(color='blue', icon='map-marker')
    ).add_to(group)

group.add_to(m)

folium.LayerControl().add_to(m)

m
```
