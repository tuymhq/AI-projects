```python
import osmnx as ox
import networkx as nx
import folium

place = "District 1, Ho Chi Minh City, Vietnam"
G = ox.graph_from_place(place, network_type='drive')

coord_start = (10.7796, 106.7000)
coord_end = (10.7720, 106.6940)

start = ox.distance.nearest_nodes(G, coord_start[1], coord_start[0])
end = ox.distance.nearest_nodes(G, coord_end[1], coord_end[0])

path1 = nx.dijkstra_path(G, start, end, weight='length')
path2 = nx.astar_path(G, start, end, weight='length')

len1 = nx.dijkstra_path_length(G, start, end, weight='length')
len2 = nx.astar_path_length(G, start, end, weight='length')

print("Dijkstra:", round(len1,0), "m")
print("A*:", round(len2,0), "m")

coords1 = [(G.nodes[n]['y'], G.nodes[n]['x']) for n in path1]
coords2 = [(G.nodes[n]['y'], G.nodes[n]['x']) for n in path2]

center = [(coord_start[0]+coord_end[0])/2, (coord_start[1]+coord_end[1])/2]
m = folium.Map(location=center, zoom_start=15)

folium.PolyLine(coords1, color='red').add_to(m)
folium.PolyLine(coords2, color='blue').add_to(m)

folium.Marker(coord_start, popup="Start").add_to(m)
folium.Marker(coord_end, popup="End").add_to(m)

m
```

**Nhận xét:**
Hai thuật toán đều tìm được đường đi ngắn nhất gần giống nhau. Tuy nhiên, A* có xu hướng chạy nhanh hơn so với Dijkstra vì nó ưu tiên đi theo hướng gần đích.

**So sánh:**
Dijkstra:
- Ưu điểm: luôn đảm bảo tìm được đường ngắn nhất
- Nhược điểm: xử lý chậm hơn khi dữ liệu lớn

A*:
- Ưu điểm: nhanh hơn do có hướng tìm kiếm rõ ràng
- Nhược điểm: cần ước lượng khoảng cách

Kết luận:
A* phù hợp khi cần tốc độ, còn Dijkstra phù hợp khi cần độ chính xác ổn định.
