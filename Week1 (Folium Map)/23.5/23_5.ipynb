```python
import folium

center = [10.775, 106.700]

m = folium.Map(location=center, zoom_start=12)

folium.Circle(location=center, radius=3000, color='green', fill=True).add_to(m)
folium.Circle(location=center, radius=5000, color='orange', fill=True).add_to(m)
folium.Circle(location=center, radius=10000, color='red', fill=True).add_to(m)

folium.Marker(center, popup="Trung tam").add_to(m)

customers = [
    [10.780, 106.710],
    [10.800, 106.730],
    [10.820, 106.750]
]

for c in customers:
    folium.Marker(c).add_to(m)

m
```

**Đánh giá:**
- 3 km: gần, giao nhanh, chi phí thấp  
- 5 km: phục vụ ổn định  
- 10 km: xa, giao lâu hơn  

**Đề xuất:**
- Nên tập trung trong phạm vi 5 km  
- Vùng xa hơn có thể mở thêm điểm nhỏ hoặc hỗ trợ giao hàng
