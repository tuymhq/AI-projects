```python
import folium
import math

customers = [
    ("A", 10.7790, 106.6980),
    ("B", 10.7730, 106.7050),
    ("C", 10.7820, 106.6920),
    ("D", 10.7700, 106.6950)
]

drivers = [
    ("1", 10.7800, 106.6990),
    ("2", 10.7720, 106.7100),
    ("3", 10.7750, 106.7010),
    ("4", 10.7830, 106.6900)
]

def dist(a, b):
    return math.sqrt((a[1]-b[1])**2 + (a[2]-b[2])**2)

assign = []
free = drivers.copy()

for c in customers:
    best = free[0]
    for d in free:
        if dist(c,d) < dist(c,best):
            best = d
    assign.append((c,best))
    free.remove(best)

m = folium.Map(location=[10.775,106.700], zoom_start=14)

for c in customers:
    folium.Marker([c[1],c[2]], popup="Khach " + c[0], icon=folium.Icon(color="red")).add_to(m)

for d in drivers:
    folium.Marker([d[1],d[2]], popup="Xe " + d[0], icon=folium.Icon(color="blue")).add_to(m)

for c,d in assign:
    folium.PolyLine([[c[1],c[2]],[d[1],d[2]]], popup=c[0] + " -> " + d[0]).add_to(m)

m
```
