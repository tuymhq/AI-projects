```python
import folium
import time
from geopy.geocoders import Nominatim
from geopy.distance import geodesic

center = (10.765451, 106.679770)

m = folium.Map(location=center, zoom_start=13)

places = [
    "Ben Thanh Market, Ho Chi Minh City",
    "Independence Palace, Ho Chi Minh City",
    "Landmark 81, Ho Chi Minh City",
    "Tan Son Nhat Airport, Ho Chi Minh City",
    "Bitexco Tower, Ho Chi Minh City",
    "Saigon Zoo, Ho Chi Minh City",
    "War Remnants Museum, Ho Chi Minh City",
    "Van Hanh Mall, Ho Chi Minh City",
    "Tao Dan Park, Ho Chi Minh City",
    "Ben Nha Rong, Ho Chi Minh City"
]

geo = Nominatim(user_agent="test")

folium.Marker(center, popup="Trung tam").add_to(m)

for addr in places:
    loc = geo.geocode(addr)
    if loc:
        point = (loc.latitude, loc.longitude)
        dist = round(geodesic(center, point).km, 2)

        folium.Marker(point, popup=str(dist) + " km").add_to(m)
        folium.PolyLine([center, point]).add_to(m)

        time.sleep(1)

m
```
