```python
!pip install osmnx
import osmnx as ox
import folium

place = "District 1, Ho Chi Minh City, Vietnam"

G = ox.graph_from_place(place, network_type='drive')

ox.plot_graph(G)

stats = ox.basic_stats(G)

print("So nut:", stats['n'])
print("So duong:", stats['m'])
print("Do dai trung binh:", round(stats['street_length_avg']/1000, 2), "km")
print("Tong do dai:", round(stats['street_length_total']/1000, 2), "km")

m = folium.Map(location=[10.775, 106.700], zoom_start=14)

nodes, edges = ox.graph_to_gdfs(G)

for _, row in edges.iterrows():
    coords = [(y, x) for x, y in row.geometry.coords]
    folium.PolyLine(coords).add_to(m)

m
```

  **Nhận xét**

*   Khu vực có nhiều nút giao → giao thông phức tạp
*   Đường dài trung bình nhỏ → mạng lưới dày
*   Mật độ đường cao → di chuyển thuận tiện hơn

  **Vai trò**

*   Giúp tìm đường nhanh hơn
*   Hỗ trợ giao hàng, vận chuyển
*   Giúp quy hoạch giao thông tốt hơn
*   Dễ quan sát và phân tích trên bản đồ
