```python
import folium
import numpy as np
from sklearn.cluster import KMeans

khach_hang = {
    'C1': (10.7824, 106.6958), 'C2': (10.7876, 106.7051), 'C3': (10.7650, 106.7020),
    'C4': (10.7790, 106.6980), 'C5': (10.7730, 106.7050), 'C6': (10.7820, 106.6920),
    'C7': (10.7700, 106.6950), 'C8': (10.7750, 106.7010), 'C9': (10.7800, 106.6990),
    'C10': (10.7720, 106.7100)
}

X = np.array(list(khach_hang.values()))

kmeans = KMeans(n_clusters=3, random_state=0)
labels = kmeans.fit_predict(X)
centers = kmeans.cluster_centers_

m = folium.Map(location=[10.775,106.700], zoom_start=15)

colors = ['red','blue','green']

for i,(name,coord) in enumerate(khach_hang.items()):
    c = labels[i]
    folium.CircleMarker(location=coord, radius=5, color=colors[c], fill=True).add_to(m)
    folium.PolyLine([coord, centers[c]], color=colors[c]).add_to(m)

for c in centers:
    folium.Marker(location=c, popup="Kho").add_to(m)

m
```

**Nhận xét:**

Dữ liệu khách hàng được chia thành 3 nhóm dựa trên vị trí địa lý.
Mỗi nhóm có một trung tâm (cluster center) được xem là vị trí đặt kho tối ưu.

Việc đặt kho gần trung tâm của mỗi cụm giúp giảm khoảng cách vận chuyển và thời gian giao hàng.
Nhờ đó có thể tối ưu chi phí và nâng cao hiệu quả phục vụ khách hàng.
