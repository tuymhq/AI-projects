```python
!pip install scikit-fuzzy
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

universe_scale = np.arange(0, 11, 0.1)

density = ctrl.Antecedent(universe_scale, 'density')
urgency = ctrl.Antecedent(universe_scale, 'urgency')
driver_load = ctrl.Antecedent(universe_scale, 'load')
traffic = ctrl.Antecedent(universe_scale, 'traffic')
profit = ctrl.Antecedent(universe_scale, 'profit')

combine = ctrl.Consequent(universe_scale, 'combine')
priority = ctrl.Consequent(universe_scale, 'priority')

density['low'] = fuzz.gaussmf(density.universe, 0, 2)
density['medium'] = fuzz.gaussmf(density.universe, 5, 1.5)
density['high'] = fuzz.gaussmf(density.universe, 10, 2)

urgency['low'] = fuzz.trapmf(urgency.universe, [0, 0, 3, 5])
urgency['medium'] = fuzz.trapmf(urgency.universe, [3, 5, 6, 8])
urgency['high'] = fuzz.trapmf(urgency.universe, [6, 8, 10, 10])

driver_load['low'] = fuzz.trimf(driver_load.universe, [0, 0, 5])
driver_load['medium'] = fuzz.trimf(driver_load.universe, [2, 5, 8])
driver_load['high'] = fuzz.trimf(driver_load.universe, [5, 10, 10])

traffic['low'] = fuzz.gaussmf(traffic.universe, 0, 2.5)
traffic['medium'] = fuzz.gaussmf(traffic.universe, 5, 2)
traffic['high'] = fuzz.gaussmf(traffic.universe, 10, 2.5)

profit['low'] = fuzz.trimf(profit.universe, [0, 0, 5])
profit['medium'] = fuzz.trimf(profit.universe, [3, 5, 7])
profit['high'] = fuzz.trimf(profit.universe, [6, 10, 10])

combine['few'] = fuzz.trapmf(combine.universe, [0, 0, 2, 4])
combine['some'] = fuzz.trapmf(combine.universe, [3, 5, 6, 8])
combine['many'] = fuzz.trapmf(combine.universe, [7, 9, 10, 10])

priority['low'] = fuzz.gaussmf(priority.universe, 0, 2)
priority['medium'] = fuzz.gaussmf(priority.universe, 5, 2)
priority['high'] = fuzz.gaussmf(priority.universe, 10, 2)

rule1 = ctrl.Rule(density['high'] & driver_load['low'] & traffic['low'], combine['many'])
rule2 = ctrl.Rule(density['medium'] & traffic['high'] & urgency['medium'], combine['some'])
rule3 = ctrl.Rule(driver_load['high'] & density['high'] & profit['medium'], combine['some'])
rule4 = ctrl.Rule(density['low'] & urgency['high'] & traffic['medium'], combine['few'])
rule5 = ctrl.Rule(profit['high'] & urgency['high'] & traffic['high'], combine['few'])

rule6 = ctrl.Rule(urgency['high'] & profit['high'], priority['high'])
rule7 = ctrl.Rule(urgency['medium'] & traffic['medium'], priority['medium'])
rule8 = ctrl.Rule(urgency['low'] & density['high'] & profit['low'], priority['low'])

delivery_ctrl = ctrl.ControlSystem([rule1, rule2, rule3, rule4, rule5, rule6, rule7, rule8])
delivery_sim = ctrl.ControlSystemSimulation(delivery_ctrl)

delivery_sim.input['density'] = 8
delivery_sim.input['urgency'] = 5
delivery_sim.input['load'] = 2
delivery_sim.input['traffic'] = 4
delivery_sim.input['profit'] = 5

delivery_sim.compute()

out_combine = delivery_sim.output['combine']
out_priority = delivery_sim.output['priority']

print(f"Ket qua so luong don ket hop: {out_combine:.2f}")
print(f"Ket qua muc do uu tien: {out_priority:.2f}")
```
